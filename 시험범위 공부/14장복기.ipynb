```python
#arr_A와 arr_B를 중첩 리스트를 이용하여 생성하라

import numpy as np

arr_A=np.arange(1,7).reshape((2,3))
arr_B=np.arange(10,70,10).reshape((2,3))

sum_arr=arr_A+arr_B
div_arr=sum_arr/2

print(div_arr)
```
