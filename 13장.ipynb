```python
arr=np.array([1,2,3,4])
mask=arr<3

packed=np.packbits(mask)
print(packed)

restored = np.unpackbits(packed)[:len(mask)]
print(restored)
#부울 배열 용량 줄이기 위해서 압축하고 푸는 방식 사용 가능
```

```python
array=[1,2,3,4]
array_bool=[1,4,3,2]

arr=np.array(array)
arr_b=np.array(array_bool)

print(arr&arr_b)
print(arr|arr_b)
print(arr==arr_b)
print(arr!=arr_b)
arr&arr_b
#넘파이에서는 파이썬 기본 비트 연산자 논리 연산자로도 사용 가능
```

```python
import numpy as np

# 어레이 생성
arr_C = np.arange(1, 11)

# 논리합(|)과 부정(~) 연산자를 활용한 조건식
# "3 이하이거나 8 이상"은 "~(3 초과 이고 8 미만)"과 동일하지만,
# 문제 조건에 맞춰 직관적으로 작성: (arr_C <= 3) | ~(arr_C < 8)
condition_C = (arr_C <= 3) | ~(arr_C < 8)

print("arr_C:", arr_C)
print("조건 만족 여부:", condition_C)
```
