```python
import numpy as np  #넘파이를 np로 임포트
```

```python
list_1D=[1,17,-7.5,3.14,2.71828]
arr_1D=np.array(list_1D)
```

```python
arr_1D #list_1D와 동일 항목 지닌 1차원 어레이 가리킴
```

```python
arr_1D.dtype #항목 자료형 확인 dtype
```

```python
arr_1D.ndim #어레이의 차원 할당자 ndim
#1차원 어레이의 항목은 왼->오 하나의 방향으로만 이동
#1차원 어레이는 1개의 축만 지님
```

```python
arr_1D.shape #튜플 자료형으로 자신의 모양 저장
#1차원 어레이의 모양은 길이가 1인 튜플
#arr_1D가 가리키는 1차원 어레이는 5개 항목(불변성)
```

요소가 하나인 튜플 선언시 쉼표 없으면 정수형으로 인식하며, 수학적 식에서 괄호는 우선순위 용도이기에 혼동되지 않도록 하기 위함이다.

1차원 배열과 2차원 배열이 있는 경우 (5, )는 벡터 (5, 1)은 행렬이다.

10,20,30,40,50은 일렬 벡터로 들어감 10 20 30 40 50은 세로 5행 가로 1열 구조를 가졌으며, 이는 각 5개의 데이터가 가로 1에 해당하는 특징을 지녔다고 볼 수 있다.
