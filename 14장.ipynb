Broadcasting이라는 규칙을 통해 자동 처리합니다. 작은 배열을 큰 배열의 크기에 맞춰 늘려서 연산을 가능하게 하는 방식입니다.

예를 들어 [1,2,3]과 [5]를 더한다고 한다면, [5]는 일차원 어레이이기 때문에 값을 복제한 것으로 생각해 [6,7,8]의 결과를 도출합니다.

```python
import numpy as np

A=np.array([1,2,3])
B=np.array([5])

print(A+B)
```

```python
arr=np.array([1,2,3,4])
mask=arr<3

packed=np.packbits(mask)
print(packed)

restored = np.unpackbits(packed)[:len(mask)]
print(restored)
```

```python
import numpy as np

# 어레이 생성
arr_C = np.arange(1, 11)

# 논리합(|)과 부정(~) 연산자를 활용한 조건식
# "3 이하이거나 8 이상"은 "~(3 초과 이고 8 미만)"과 동일하지만,
# 문제 조건에 맞춰 직관적으로 작성: (arr_C <= 3) | ~(arr_C < 8)
condition_C = (arr_C <= 3) | ~(arr_C < 8)

print("arr_C:", arr_C)
print("조건 만족 여부:", condition_C)
```
