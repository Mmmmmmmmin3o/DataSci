```python
import pandas as pd

# 예시 DatetimeIndex
two_days = pd.date_range('2026-03-01 00:00:00', '2026-03-02 23:00:00', freq='H')

# 원하는 범위
start = pd.Timestamp('2026-03-02 00:00:00')
end   = pd.Timestamp('2026-03-02 11:00:00')

# for문으로 필터링
selected = [t for t in two_days if start <= t <= end]
print(pd.DatetimeIndex(selected))
```

```python
start = pd.Timestamp('2026-03-02 00:00:00')
end   = pd.Timestamp('2026-03-02 11:00:00')

selected=[]

for t in two_days:
    if start<=t and t<=end:
        selected.append(t)

selected=pd.DatetimeIndex(selected)
type(selected)
```
