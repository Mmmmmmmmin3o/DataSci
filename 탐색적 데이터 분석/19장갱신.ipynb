```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns
```

```python
pd.set_option('display.precision',6)
```

```python
penguins=sns.load_dataset('penguins')
penguins.head()
```

```python
penguins.info()
```

```python
penguins=penguins.dropna()
penguins.info()
```

```python
#그래프 스타일 지정
plt.rc('figure',figsize=(6.4,4.8))
#그래프 가로 세로 크기(인치 기준)
plt.rc('font',size=10)
#기본 글자 크기 설정
plt.rc('axes', labelsize=10, titlesize=12)
#축 이름 크기/그래프 제목 크기 설정
plt.rc('legend', fontsize=10)
#범례 글자 크기 설정
plt.rc('xtick', labelsize=10)
#x축 눈금 레이블 크기 설정
plt.rc('ytick',labelsize=10)
#y축 눈금 레이블 크기 설정
```

```python
#그래프 그리기 세가지 방법
#1. matplotlib 사용
#pyplot 함수형, 객체지향 OO
#서브플롯 여러개 다룰 때 주로 사용
#2. 판다스 데이터 프레임/시리즈 메서드
#판다스가 내부적으로 matplotlib 호출
#코드가 짧고 직관적
#3. seaborn 라이브러리 활용
#통계 시각화 특화 라이브러리.
#내부적으로 matplotlib사용하지만 정돈됨

#세 방식 모두 matplotlib를 거침
#plt.xlabel(), plt.title(), plt.show() 다 사용 가능
```

```python
#pyplot 함수형 히스토그램
plt.hist(penguins['bill_length_mm'], bins=20, edgecolor='black')

plt.xlabel('Bill Length (mm)')
plt.ylabel('Frequency')
plt.title("Bill Length (plt.hist)")
plt.show()
```

```python
#판다스 데이터 프레임 히스토그램 메서드
penguins['bill_length_mm'].hist(bins=20, edgecolor='black', grid=False)

plt.xlabel('bill Length (mm)')
plt.ylabel('Frequency')
plt.title('Bill Length (df.hist)')
plt.show()
```

```python
#sns 라이브러리 히스토그램 작성
sns.histplot(data=penguins, x='bill_length_mm',bins=20, hue='species')

# plt.xlabel('bill Length (mm)')
# plt.ylabel('Frequency')
#없어도 자동 지정해줌
plt.title('bill Length by Species (sns.histplot)')
plt.show()
```
