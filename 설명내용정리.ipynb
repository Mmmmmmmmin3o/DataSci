```python
#파이썬은 함수를 인자로 사용가능

def g():
    return 3

def h(f):
    return f()

print(h(g))
```

```python
#파이썬에서 클래스 생성

class A:
    pass

a=A()

#객체는 클래스에의해 만들어진 값
#함수도 함수 클래스의 객체

#값에 자료형 주어지는건 그냥 주어진것
```

```python
class B:
    def __init__(self,a):
        self.len=a
    def length(self):
        return self.len
```

```python
b=B(3)
```

```python
b.length()

#파이썬에선 클래스와 리스트가 동일?
```
