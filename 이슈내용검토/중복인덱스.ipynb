# 임의의 데이터로 확인하는 인덱스 중복 여부

```python
import pandas as pd
import numpy as np

series1 = pd.Series([4, 5, 6, 3, 1], name="Mango")
series2 = pd.Series([4, 5, 6, 3, 1], name="Mango")
series3 = pd.Series([2, 3, 5, 2, 7], name="Banana")

fruits1 = pd.concat([series1, series2, series3], axis=1)
fruits1
```

```python
#컬럼 중복 확인
fruits1.columns.duplicated(keep=False)
```

```python
#중복 컬럼 출력
dup_cols=fruits1.iloc[:,fruits1.columns.duplicated(keep=False)]
print(dup_cols)
```

```python
#중복 컬럼(T 사용해 행렬 전치) 출력
fruits1.T.duplicated()
```

```python
#중복 컬럼 삭제 후 확인
fruits1=fruits1.loc[:,~fruits1.T.duplicated()]
fruits1
```

# housing.csv의 중복 여부

```python
data_url = 'https://raw.githubusercontent.com/codingalzi/code-workout-datasci/refs/heads/master/data/'
```

```python
housing = pd.read_csv(data_url+'housing.csv')
```

```python
#중복값 존재하지 않아 강제 생성
housing['total_rooms_dup'] = housing['total_rooms']
housing['total_rooms_dup2']=housing['latitude']
housing.rename(columns={'total_rooms_dup': 'total_rooms'}, inplace=True)
housing.rename(columns={'total_rooms_dup2': 'total_rooms'}, inplace=True)
```

```python
#중복 컬럼 생성 확인
print(housing.loc[:,'total_rooms'])
```

```python
#원본 데이터 확인
housing
```

```python
#열 기준 중복 데이터값 기준 삭제
housing_boolarr=housing.T.duplicated()
housing=housing.loc[:, ~housing_boolarr]
housing

#행기준 중복 데이터 값 삭제
# housing=housing.drop_duplicates()
# housing
```

```python
#중복 인덱스 기준 삭제
housing_boolarr=housing.columns.duplicated()
housing=housing.loc[:,~housing_boolarr]
housing
```

```python
#중복 값과 인덱스 기준 삭제
housing_boolarr=housing.T.duplicated()&housing.columns.duplicated()
housing=housing.loc[:,~housing_boolarr]
housing
```

```python
boolean=housing.columns.duplicated()
```

```python
housing.columns.values[boolean] = 'check'
```

```python
housing
```

```python
cols = pd.Series(housing.columns)
print(cols[cols.duplicated()])
for dup in cols[cols.duplicated()].unique():
  print("중복 데이터",dup)
  Idx=cols[cols==dup].index
  print("인덱스",Idx)
  new_cols = []

  for i in range(len(Idx)):
      if i == 0:
          new_cols.append(dup)          # 첫 번째는 그대로
          print("중복컬럼 0",new_cols[i])
      else:
          new_cols.append(f"{dup}_{i}") # 나머지는 번호 붙임
          print("중복컬럼",i,new_cols[i])

  cols[Idx] = new_cols

housing.columns = cols
housing
```

```python
print(f"인덱스 중복 없는가? : {housing.columns.is_unique}")
```
