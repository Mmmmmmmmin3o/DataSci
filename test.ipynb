```python
#numpy : 수치 연산
#pandas : 데이터 프레임 기반 데이터 처리
#matplotlib.pyplot : 데이터 시각화
#seaborn : 통계 기반 데이터 시각화

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.set_option('display.precision', 6) #소수점 이하 6자리까지만 출력
```

```python
data_url = 'https://raw.githubusercontent.com/codingalzi/code-workout-datasci/refs/heads/master/data/'
```

```python
iris_df = pd.read_csv(data_url + 'iris.csv')
iris_df
```

```python
iris_df.info()
```

```python
pl=iris_df['petal.length']
pl.mean()
```

```python
sw=iris_df['sepal.width']
sw.median()
```

```python
pw=iris_df['petal.width']
pw.mode()[0]
```

```python
sl=iris_df['sepal.length']
Q1=sl.quantile(0.25)
Q3=sl.quantile(0.75)
IQR=Q3-Q1

print(f"제1사분위 {Q1}, 제3사분위 {Q3}, 사분범위 {IQR}")
```

```python
iris_df.boxplot(
    column=['sepal.width'],
    grid=False,
    flierprops=dict(marker='o',markerfacecolor='red',markersize=5)
)
```

```python
범위=pw.max()-pw.min()
분산=pw.var(ddof=0)
표준편차=pw.std(ddof=0)

print(f"범위 {범위}, 분산 {분산} 표준편차 {표준편차} ")
```

```python
scale=(pl-pl.min())/(pl.max()-pl.min())
iris_df['petal_length_scaled']=scale
iris_df[['petal.length','petal_length_scaled']].head()
```

```python
sepal_z=(sw-sw.mean())/sw.std(ddof=0)
iris_df['sepal_width_z']=sepal_z
iris_df[['sepal.width','sepal_width_z']].head()
```

```python
Q1 = sw.quantile(0.25)
Q3 = sw.quantile(0.75)
IQR = Q3 - Q1
lowerCase=Q1-1.5*IQR
upperCase=Q3+1.5*IQR

outliers=iris_df[(sw<lowerCase)|(sw>upperCase)]
print(f"이상치 개수 : {len(outliers)}")
outliers
```
