### 머신러닝과 사이킷런 라이브러리
* 머신러닝 모델은 입출력 데이터 사이의 관계를 학습, 새로운 데이터에 대해 예측하거나 판단하는 알고리즘

* 머신러닝 수업에서는 과거에 사용한 오래된 기술을 다룰 것. 그렇다고 전통적인 방식이 무조건 좋지 않다는 것은 아님

* 딥러닝과 같은 경우에는 아주 많은 데이터를 필요로 한다. (후에 2학기에는 딥러닝 중심 학습, 간단한 LLM 제작 할 것)
