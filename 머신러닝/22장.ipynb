```python
training_columns=['bill_length_mm',
                  "bill_depth_mm",
                  "flipper_length_mm",
                  "body_mass_g",
                  "species"]

penguins_training=penguins[training_columns].copy()
```

```python
X_bl=penguins_training.loc[:,"bill_length_mm"]
X_bl_Scaled=(X_bl-X_bl.mean())/X_bl.std()
```
